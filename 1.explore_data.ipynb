```python
card_trans = pd.read_csv('../assets/cc_txn.csv')
card_trans["txn_dt"] = pd.to_datetime(card_trans["txn_dt"], errors="coerce")
card_trans["year_month"] = card_trans["txn_dt"].dt.to_period("M").astype(str)
card_trans["month"] = card_trans["txn_dt"].dt.month
print(card_trans.shape)
card_trans.head()
```

```python
print(card_trans['txn_dt'].min())
print(card_trans['txn_dt'].max())
```

```python
card_type_agg = pd.DataFrame(card_trans.groupby(['user_id', 'year_month', 'mcc', 'card_no']).agg(sum_amt=('txn_amt', 'sum'), avg_amt=('txn_amt', 'mean'), cnt_trans=("user_id","count"))).reset_index()
card_type_agg
```

```python
card_agg = pd.DataFrame(card_trans.groupby(['user_id', 'year_month', 'card_no']).agg(sum_amt=('txn_amt', 'sum'), avg_amt=('txn_amt', 'mean'), cnt_trans=("user_id","count"))).reset_index()
card_agg
```

```python
card_agg = card_agg.merge(card, on=['user_id', 'card_no'], how='left')
card_agg
```

```python
'''
0 -> เกินวงเงิน
1 -> ไม่เกินวงเงิน
'''
card_agg['utilization_ratio'] = (card_agg['sum_amt']) / card_agg['cr_lmt_amt'] *100
card_agg['tag'] = np.where(card_agg['utilization_ratio']>100, 0, 1)
card_agg
```

```python
card_agg['tag'].value_counts()
```
